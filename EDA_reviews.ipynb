```python
def basic_clean(text):
    if pd.isna(text):
        return ''

    text = text.lower()
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()

    return text

df['clean_text'] = df['text'].apply(basic_clean)
```

```python
df['clean_word_count'] = df['clean_text'].str.split().str.len()
```

```python
def remove_rating_block(text):
    if pd.isna(text):
        return ''

    text = text.lower()
    text = re.sub(r'\bmore pics\b', '', text)
    text = re.sub(r'\b(overall|story|animation|sound|character|enjoyment)\s+\d+\b', '', text, flags=re.IGNORECASE)
    text = re.sub(r'[^a-zA-Z0-9\s!?]', ' ', text)
    text = re.sub(r'\s+', ' ', text)
    text = text.strip()

    return text

df['cleaner_text'] = df['text'].apply(remove_rating_block)
```
