```python
import pandas as pd
import numpy as np

from sklearn.svm import LinearSVC, SVC
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score
```

```python
df = pd.read_csv('../After_EDA_data/Light_text.csv')
```

```python
def classification(text):
    if text >= 5:
        return 'Good'
    return "Bad"
```

```python
df['target'] = df['score'].apply(classification)
```

```python
df['text'] = df['text'].drop_duplicates()
```

```python
df = df.dropna()
```

```python
df.info()
```

```python
df_majority = df[df['target'] == 'Good']
df_minority = df[df['target']== 'Bad']
```

```python
df_reduced = df_majority.sample(n=len(df_minority), random_state=42)
```

```python
df_balanced = pd.concat([df_reduced, df_minority]).sample(frac=1, random_state=42)
```

```python
df_balanced
```

```python
df_balanced = df_balanced.reset_index(drop=True)
```

```python
df_balanced
```

```python
df_balanced['text'].duplicated().sum()
```

```python
df_balanced.info()
```

```python
x = df_balanced['text']
y = df_balanced['target']
```

```python
le = LabelEncoder()
y_encoded = le.fit_transform(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    x, y_encoded, test_size=0.3, stratify=y_encoded, random_state=42
)
```

```python
tfidf = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2)
)
```

```python
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)
```

```python
svc = LinearSVC()

svc.fit(X_train_tfidf, y_train)
```

```python
y_pred = svc.predict(X_test_tfidf)
y_score = svc.decision_function(X_test_tfidf)
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(confusion_matrix(y_test, y_pred))
```

```python
print(roc_auc_score(y_test, y_score))
```
